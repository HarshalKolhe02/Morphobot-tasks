```python
from matplotlib import pyplot as plt
```

```python
import numpy as np
import math
```

```python
u_t=0
t=0
prev_er=0
inte=0
T_set=24
k_p=5
k_i=5
k_d=0.1
pow=[]
temp=[]
time=[]
er_arr=[]
```

```python
while(t<500):
  T= 35+(0.02*t)-(0.05*(u_t))
  er=T-T_set
  inte+=er*0.1
  dev=(er-prev_er)/0.1
  prev_er=er
  P=k_p*er
  I=k_i*inte
  D=k_d*dev
  u_t=P+I+D
  pow.append(round(u_t,1))
  temp.append(round(T,2))
  time.append(t)
  t+=0.1
  er_arr.append(round(er,3))
```

```python
print(len(time))
print(len(temp))
```

```python
pow=np.array(pow)
time=np.array(time)
temp=np.array(temp)
er_arr=np.array(er_arr)


plt.plot(time,temp)
plt.xlabel("Time")
plt.ylabel("Temperature")

plt.show()

```

```python
plt.plot(time,pow)
plt.xlabel("Time")
plt.ylabel("Power")
plt.show()
```

```python
plt.xlabel("Time")
plt.ylabel("Error")
plt.plot(time,er_arr)
plt.show()
```
