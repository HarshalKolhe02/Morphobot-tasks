```python
from matplotlib import pyplot as plt
```

```python
import numpy as np
import math
```

```python
u_t=0
t=0
prev_er=0
inte=0
H_set=5
k_p=5
k_i=10
k_d=0
flow=[]
level=[]
time=[]
er_arr=[]
```

```python
while((t<500)and(u_t>=0)):
  H=2+(0.01*t)+(0.05*(u_t))
  er=H_set-H
  inte+=er*0.1
  dev=(er-prev_er)/0.1
  prev_er=er
  P=k_p*er
  I=k_i*inte
  D=k_d*dev
  u_t=P+I+D
  flow.append(round(u_t,3))
  level.append(round(H,3))
  time.append(t)
  t+=0.1
  er_arr.append(round(er,3))
```

```python
flow=np.array(flow)
level=np.array(level)
time=np.array(time)
print(level)
print(len(level))
#er_arr=np.array(er_arr)
plt.plot(time,level)

plt.axhline(H_set, linestyle="--", label="Set Level")
plt.legend()
plt.xlabel("Time")
plt.ylabel("Level")
plt.show()
#print(flow)
#plt.plot(time,er_arr)
#plt.show()
```
