```python
from matplotlib import pyplot as plt
import numpy as np
import math
```

```python
u_t=0
t=0
prev_er=0
inte=0
H_set=10
k_p=5
k_i=5
k_d=0.1
alt=[]
thr=[]
time=[]
er_arr=[]
```

```python
while((t<500)and(u_t>=0)):
  H=2-(0.002*t)+(0.03*(u_t))
  er=H_set-H
  inte+=er*0.1
  dev=(er-prev_er)/0.1
  prev_er=er
  P=k_p*er
  I=k_i*inte
  D=k_d*dev
  u_t=P+I+D
  thr.append(round(u_t,3))
  alt.append(round(H,3))
  time.append(t)
  t+=0.1
  er_arr.append(round(er,3))
```

```python
thr=np.array(thr)
alt=np.array(alt)
time=np.array(time)
#print(level)
#print(len(level))
er_arr=np.array(er_arr)
plt.plot(time,alt)

plt.axhline(H_set, linestyle="--", label="Desired Alt")
plt.legend()
plt.xlabel("Time")
plt.ylabel("Altitude")
plt.show()
```

```python
plt.plot(time,thr)
plt.xlabel("Time")
plt.ylabel("Throttle")
plt.show()
```

```python
plt.plot(time,er_arr)
plt.xlabel("Time")
plt.ylabel("Error")
plt.show()
```
